```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_csv("superstore_cleaned.csv")
df["order_date"] = pd.to_datetime(df["order_date"])
df["ship_date"] = pd.to_datetime(df["ship_date"])
print(df.shape)
```

```python
engine = create_engine("postgresql://postgres:Zxcvbnm,./12@localhost:5432/Sales_db")

# Test connection
with engine.connect() as conn:
    print("Connection successful!")
```

```python
df.to_sql(
    name="sales_data",
    con=engine,
    if_exists="replace",      
    index=False,              
    chunksize=500             
)
print("Data loaded successfully!")
```

```python
import pandas as pd

df_check = pd.read_sql("SELECT * FROM sales_data LIMIT 5", engine)
df_check
```

```python
df_count = pd.read_sql("SELECT COUNT(*) FROM sales_data", engine)
print(df_count)
```
